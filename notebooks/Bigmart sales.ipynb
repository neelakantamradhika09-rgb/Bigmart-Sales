Step 24: Seperate Features and Target

```python
X=train.drop('Item_Outlet_Sales', axis=1)
y=train['Item_Outlet_Sales']
print("X shape:", X.shape)
print("y shape:", y.shape)
```

The target variable Item_Outlet_Sales is separated from the training dataset.
X contains the input features that will be used for model training, while y contains the target variable that the model needs to predict.
The shapes of X and y are displayed to verify that the data has been separated correctly.

Step 25:Train Validation Split

```python
from sklearn.model_selection import train_test_split
X_train, X_valid, y_train, y_valid=train_test_split(
    X,y,test_size=0.20,random_state=42)
print("X_train:", X_train.shape)
print("X_valid:", X_valid.shape)
print("y_train:", y_train.shape)
print("y_valid:", y_valid.shape)
```

The dataset is divided into training and validation sets using train_test_split().
80% of the data is used for training the models, while the remaining 20% is used for validation.
The random_state is set to 42 to ensure the same split is obtained each time the code is executed.
The shapes of all four sets are displayed to verify the split.

Step 26: Encode Categorical Variables

```python
from sklearn.preprocessing import OneHotEncoder
categorical_columns=X_train.select_dtypes(
    include=['object']).columns
numerical_columns=X_train.select_dtypes(
    include=['int64', 'float64']).columns
encoder=OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False)
X_train_cat=encoder.fit_transform(
    X_train[categorical_columns])
X_valid_cat = encoder.transform(
    X_valid[categorical_columns])
print("Categorical variables encoded successfully")
```

Categorical and numerical columns are identified separately from the training data.
OneHotEncoder is used to convert the categorical variables into numerical form so that they can be used by machine learning models.
The encoder is fitted on the training data and then applied to the validation data.
The handle_unknown='ignore' option prevents errors when the validation data contains a category that was not present in the training data.
The encoded categorical data is stored separately for further preprocessing.

Step 27: Prepare Final Numerical Data

```python
import numpy as np
X_train_processed=np.hstack([
    X_train[numerical_columns].values,
    X_train_cat])
X_valid_processed=np.hstack([
    X_valid[numerical_columns].values,
    X_valid_cat])
print("X_train processed shape:", X_train_processed.shape)
print("X_valid processed shape:", X_valid_processed.shape)
```

The numerical columns and encoded categorical columns are combined to create the processed training and validation datasets.
The numpy hstack() function is used to join these columns horizontally.
The processed datasets are converted into a numerical format that can be directly given to machine learning models.
The shapes of the processed datasets are displayed to verify that the preprocessing was completed correctly.

Step 28: Import Libraries

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

The Linear Regression model and evaluation metrics are imported for model training and performance evaluation.
LinearRegression is used to build a regression model for predicting Item_Outlet_Sales.
mean_squared_error is used to measure the prediction error, while r2_score is used to evaluate how well the model explains the variation in the target variable.
NumPy is imported for numerical calculations.

Model 1: Linear Regression

```python
linear_model=LinearRegression()
linear_model.fit(X_train_processed, y_train)
linear_predictions=linear_model.predict(X_valid_processed)
linear_mae=mean_absolute_error(y_valid, linear_predictions)
linear_mse=mean_squared_error(y_valid, linear_predictions)
linear_rmse=np.sqrt(linear_mse)
linear_r2=r2_score(y_valid, linear_predictions)
print("Linear Regression Results")
print("MAE      :",round(linear_mae,3))
print("MSE      :", round(linear_mse,3))
print("RMSE     :", round(linear_rmse,3))
print("R2 Score :", round(linear_r2,3))
```

The Linear Regression model is trained using the processed training data and used to predict Item_Outlet_Sales for the validation data.
The model performance is evaluated using MAE, MSE, RMSE, and R2 Score.
MAE shows the average absolute difference between actual and predicted values.
MSE measures the average squared prediction error, while RMSE represents the square root of MSE.
R2 Score indicates how well the model explains the variation in the target variable.
The results are rounded to three decimal places for easier comparison with other regression models.

Model 2: RandomForest

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(
    n_estimators=200,max_depth=10,random_state=42,n_jobs=-1)
rf_model.fit(X_train_processed, y_train)
rf_predictions=rf_model.predict(X_valid_processed)
rf_mae=mean_absolute_error(y_valid, rf_predictions)
rf_mse=mean_squared_error(y_valid, rf_predictions)
rf_rmse=np.sqrt(rf_mse)
rf_r2=r2_score(y_valid, rf_predictions)
print("Random Forest Regression Results")
print("MAE      :",round (rf_mae,3))
print("MSE      :",round (rf_mse,3))
print("RMSE     :", round(rf_rmse,3))
print("R2 Score :", round(rf_r2,3))
```

The Random Forest Regressor is trained using the processed training data and used to predict Item_Outlet_Sales for the validation data.
The model performance is evaluated using MAE, MSE, RMSE, and R2 Score.
Random Forest uses multiple decision trees to make predictions and can capture non-linear relationships in the data.
The calculated evaluation metrics help compare the Random Forest model with the Linear Regression model.

Model 3: Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor
gb_model = GradientBoostingRegressor(
    n_estimators=200,learning_rate=0.05,max_depth=4,random_state=42)
gb_model.fit(X_train_processed, y_train)
gb_predictions=gb_model.predict(X_valid_processed)
gb_mae=mean_absolute_error(y_valid, gb_predictions)
gb_mse=mean_squared_error(y_valid, gb_predictions)
gb_rmse=np.sqrt(gb_mse)
gb_r2=r2_score(y_valid, gb_predictions)
print("Gradient Boosting Regression Results")
print("MAE      :", round(gb_mae,3))
print("MSE      :", round(gb_mse,3))
print("RMSE     :", round(gb_rmse,3))
print("R2 Score :", round(gb_r2,3))
```

The Gradient Boosting Regressor is trained using the processed training data and used to predict Item_Outlet_Sales for the validation data.
The model performance is evaluated using MAE, MSE, RMSE, and R2 Score.
Gradient Boosting builds multiple decision trees sequentially, where each new tree improves the errors made by the previous trees.
The evaluation results can be compared with Linear Regression and Random Forest to identify the better-performing regression model.

Model 4: Lasso Regression

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
lasso_model=Lasso(
    alpha=1.0,max_iter=10000)
lasso_model.fit(
    X_train_processed,y_train)
lasso_predictions=lasso_model.predict(X_valid_processed)
lasso_mae=mean_absolute_error(y_valid, lasso_predictions)
lasso_mse=mean_squared_error(y_valid, lasso_predictions)
lasso_rmse=np.sqrt(lasso_mse)
lasso_r2=r2_score(y_valid, lasso_predictions)
print("Lasso Regression Results")
print("MAE      :", round(lasso_mae,3))
print("MSE      :", round(lasso_mse,3))
print("RMSE     :", round(lasso_rmse,3))
print("R2 Score :", round(lasso_r2,3))
```

The Lasso Regression model is trained using the processed training data and used to predict Item_Outlet_Sales for the validation data.
Lasso Regression applies regularization to reduce the effect of less important features and can help prevent overfitting.
The model performance is evaluated using MAE, MSE, RMSE, and R2 Score.
The evaluation results can be compared with the other regression models to identify the better-performing model.

Model 5: Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
tree_model=DecisionTreeRegressor(
    max_depth=8,
    random_state=42)
tree_model.fit(
    X_train_processed,y_train)
tree_predictions=tree_model.predict(X_valid_processed)
tree_mae=mean_absolute_error(y_valid, tree_predictions)
tree_mse=mean_squared_error(y_valid, tree_predictions)
tree_rmse=np.sqrt(tree_mse)
tree_r2=r2_score(y_valid, tree_predictions)
print("Decision Tree Regression Results")
print("MAE      :", round(tree_mae,3))
print("MSE      :", round(tree_mse,3))
print("RMSE     :", round(tree_rmse,3))
print("R2 Score :", round(tree_r2,3))
```

The Decision Tree Regression model is trained using the processed training data and used to predict Item_Outlet_Sales for the validation data.
The max_depth parameter controls the maximum depth of the decision tree and helps control model complexity.
The model performance is evaluated using MAE, MSE, RMSE, and R2 Score.
The evaluation results can be compared with the other regression models to identify the better-performing model.
